```python
# %% [code]
# =============================================================================
# IMPORT LIBRARIES & SETUP
# =============================================================================
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_style('whitegrid')

import warnings
warnings.filterwarnings('ignore')

import sklearn.metrics as skm
import sklearn.model_selection as skms
import sklearn.preprocessing as skp
from sklearn.decomposition import PCA
from sklearn.preprocessing import MinMaxScaler

import librosa
import librosa.display as lplt
import IPython

import tensorflow as tf
import keras as k

# Setting up the random seeds for reproducibility
seed = 12
np.random.seed(seed)
tf.random.set_seed(seed)

# =============================================================================
# Doing AUDIO PROCESSING & VISUALIZATION FUNCTIONS
# =============================================================================
def load_and_trim_audio(audio_fp):
    """Doing Loading an audio file and trim silent edges."""
    audio_data, sr = librosa.load(audio_fp)
    audio_data, _ = librosa.effects.trim(audio_data)
    return audio_data, sr

def play_audio(audio_data, sr):
    """Returning the IPython Audio object for playback."""
    return IPython.display.Audio(audio_data, rate=sr)

def plot_waveform(audio_data, sr):
    """Plotting the audio waveform."""
    plt.figure(figsize=(15, 5))
    librosa.display.waveshow(audio_data, sr=sr)
    plt.xlabel("Time (s)")
    plt.ylabel("Amplitude")
    plt.title("Waveform")
    plt.show()

def plot_stft(audio_data, sr, n_fft=2048, hop_length=512, cmap='cool'):
    """Plotting the log-scaled spectrogram from STFT."""
    stft = librosa.stft(audio_data, n_fft=n_fft, hop_length=hop_length)
    stft_db = librosa.amplitude_to_db(np.abs(stft), ref=np.max)
    plt.figure(figsize=(12, 4))
    librosa.display.specshow(stft_db, sr=sr, hop_length=hop_length,
                             x_axis='time', y_axis='log', cmap=cmap)
    plt.colorbar(format='%+2.0f dB')
    plt.title("Log-scaled Spectrogram (dB)")
    plt.tight_layout()
    plt.show()

def plot_zoomed_wave(audio_data, start=1000, end=1200):
    """Plotting the zoomed section of the audio waveform."""
    plt.figure(figsize=(16, 4))
    plt.plot(audio_data[start:end])
    plt.title("Zoomed Audio Waveform")
    plt.show()

def plot_mel_spectrogram(audio_data, sr, n_fft=2048, hop_length=512, n_mels=128, cmap='cool'):
    """Plotting the Mel spectrogram in decibel scale."""
    mel_spec = librosa.feature.melspectrogram(y=audio_data, sr=sr, n_fft=n_fft,
                                              hop_length=hop_length, n_mels=n_mels)
    mel_spec_db = librosa.amplitude_to_db(mel_spec, ref=np.max)
    plt.figure(figsize=(16, 6))
    librosa.display.specshow(mel_spec_db, sr=sr, hop_length=hop_length,
                             x_axis='time', y_axis='mel', cmap=cmap)
    plt.colorbar(format='%+2.0f dB')
    plt.title("Mel Spectrogram (dB)")
    plt.tight_layout()
    plt.show()

def plot_chroma(audio_data, sr, hop_length=512, cmap='coolwarm'):
    """Plotting the chroma features."""
    chroma = librosa.feature.chroma_stft(y=audio_data, sr=sr, hop_length=hop_length)
    plt.figure(figsize=(16, 6))
    librosa.display.specshow(chroma, sr=sr, hop_length=hop_length,
                             x_axis='time', y_axis='chroma', cmap=cmap)
    plt.colorbar()
    plt.title("Chroma Features")
    plt.tight_layout()
    plt.show()

# =============================================================================
# Doing the FEATURE ANALYSIS & VISUALIZATION FUNCTIONS
# =============================================================================
def load_dataset(csv_path):
    """Loading the CSV dataset and printing some basic information."""
    df = pd.read_csv(csv_path)
    print("Dataset has", df.shape)
    print("Count of samples by label:")
    print(df.label.value_counts().reset_index())
    return df

def plot_correlation_heatmap(df, save_path="mean_feature_correlation_heatmap.png"):
    """Plotting a heatmap of correlations for mean-based features."""
    mean_cols = [col for col in df.columns if 'mean' in col]
    corr_matrix = df[mean_cols].corr()
    mask = np.triu(np.ones_like(corr_matrix, dtype=bool))

    plt.figure(figsize=(16, 11))
    custom_cmap = sns.diverging_palette(0, 25, s=90, l=45, n=5, as_cmap=True)
    sns.heatmap(corr_matrix, mask=mask, cmap=custom_cmap, vmax=0.3, center=0,
                square=True, linewidths=0.5, cbar_kws={"shrink": 0.5})
    plt.title("Feature Correlation Heatmap (Mean-based)", fontsize=20)
    plt.xticks(fontsize=10, rotation=45, ha='right')
    plt.yticks(fontsize=10)
    plt.tight_layout()
    plt.savefig(save_path)
    plt.show()

def plot_tempo_boxplot(df, save_path="genre_tempo_boxplot.png"):
    """Creating the boxplot of tempo distribution across genres."""
    tempo_data = df[["label", "tempo"]]
    plt.figure(figsize=(16, 8))
    sns.boxplot(data=tempo_data, x="label", y="tempo", palette="husl")
    plt.title("Tempo Distribution by Genre (BPM)", fontsize=20)
    plt.xlabel("Music Genre", fontsize=15)
    plt.ylabel("Beats Per Minute", fontsize=15)
    plt.xticks(fontsize=14, rotation=45, ha='right')
    plt.yticks(fontsize=10)
    plt.tight_layout()
    plt.savefig(save_path)
    plt.show()

def plot_pca(X_normalized, genres, save_path="pca_genre_scatterplot.png"):
    """Applying the PCA to the normalized features and plot a 2D scatter of genres."""
    pca = PCA(n_components=2)
    pca_result = pca.fit_transform(X_normalized)
    pca_df = pd.DataFrame(pca_result, columns=["PC1", "PC2"])
    pca_df['Genre'] = genres
    plt.figure(figsize=(16, 9))
    sns.scatterplot(data=pca_df, x="PC1", y="PC2", hue="Genre", alpha=0.7, s=100)
    plt.title("PCA Projection of Music Genres", fontsize=20)
    plt.xlabel("Principal Component 1", fontsize=15)
    plt.ylabel("Principal Component 2", fontsize=15)
    plt.xticks(fontsize=14)
    plt.yticks(fontsize=10)
    plt.tight_layout()
    plt.savefig(save_path)
    plt.show()

# =============================================================================
# Doing DATA PREPROCESSING FUNCTIONS
# =============================================================================
def preprocess_data(df, seed=12):
    """
    splitting the data into training, validation, and test sets, shuffle the data,
    remove any unnecessary columns, and map genre labels to indices.scales the features as well.

    """
    # Creating the genre mapping
    genre_to_index = {label: idx for idx, label in enumerate(df['label'].unique())}
    index_to_genre = {idx: label for label, idx in genre_to_index.items()}
    print("Genre to Index Mapping:\n", genre_to_index)
    print("\nIndex to Genre Mapping:\n", index_to_genre)

    # Replacing the labels with indices
    df['label'] = df['label'].map(genre_to_index)

    # Shuffling the drop non-feature columns
    df_shuffled = df.sample(frac=1, random_state=seed).reset_index(drop=True)
    df_shuffled.drop(columns=["filename", "length"], inplace=True)

    # Separating the features and target
    target = df_shuffled.pop("label")
    features = df_shuffled

    # Splitting the data into training (70%), validation (20%) and test (10%)
    X_train, X_temp, y_train, y_temp = skms.train_test_split(
        features, target, train_size=0.7, random_state=seed, stratify=target
    )
    X_val, X_test, y_val, y_test = skms.train_test_split(
        X_temp, y_temp, train_size=0.66, random_state=seed, stratify=y_temp
    )

    total = len(df_shuffled)
    print(f"Training Set: {X_train.shape[0]} samples ({round(X_train.shape[0] / total * 100)}% of total)")
    print(f"Validation Set: {X_val.shape[0]} samples ({round(X_val.shape[0] / total * 100)}% of total)")
    print(f"Test Set: {X_test.shape[0]} samples ({round(X_test.shape[0] / total * 100)}% of total)")

    print(f"Training Set - Majority Class Proportion: {y_train.value_counts(normalize=True).max() * 100:.2f}%")
    print(f"Validation Set - Majority Class Proportion: {y_val.value_counts(normalize=True).max() * 100:.2f}%")
    print(f"Test Set - Majority Class Proportion: {y_test.value_counts(normalize=True).max() * 100:.2f}%")

    # Doing Standardize features
    scaler = skp.StandardScaler()
    X_train_scaled = pd.DataFrame(scaler.fit_transform(X_train), columns=X_train.columns)
    X_val_scaled = pd.DataFrame(scaler.transform(X_val), columns=X_train.columns)
    X_test_scaled = pd.DataFrame(scaler.transform(X_test), columns=X_train.columns)

    return X_train_scaled, X_val_scaled, X_test_scaled, y_train, y_val, y_test, genre_to_index, index_to_genre

# =============================================================================
# Doing MODEL TRAINING FUNCTIONS & ARCHITECTURES
# =============================================================================
ACCURACY_THRESHOLD = 0.94

class AccuracyEarlyStop(k.callbacks.Callback):
    """
    Here When the validation accuracy threshold is reached, a custom callback will end training early.
    """
    def on_epoch_end(self, epoch, logs=None):
        if logs.get("val_accuracy") > ACCURACY_THRESHOLD:
            print(f"\nEarly stopping: Validation accuracy exceeded {ACCURACY_THRESHOLD * 100:.2f}%")
            self.model.stop_training = True

def train_and_plot(model, X_train, y_train, X_val, y_val, epochs, optimizer, batch_size=128):
    """Compile, train the model, and plot training history."""
    model.compile(optimizer=optimizer,
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])

    history = model.fit(
        X_train, y_train,
        validation_data=(X_val, y_val),
        epochs=epochs,
        batch_size=batch_size,
        callbacks=[AccuracyEarlyStop()]
    )

    best_val_acc = max(history.history["val_accuracy"])
    print(f"Highest Validation Accuracy Achieved: {best_val_acc:.4f}")

    pd.DataFrame(history.history).plot(figsize=(12, 6))
    plt.title("Training and Validation Metrics Over Epochs")
    plt.xlabel("Epoch")
    plt.ylabel("Metric Value")
    plt.grid(True)
    plt.tight_layout()
    plt.show()
    return history

def build_model_1(input_dim, num_classes):
    """Here I'm Building a simple feedforward neural network."""
    model = k.models.Sequential([
        k.layers.Dense(256, activation='relu', input_shape=(input_dim,)),
        k.layers.Dense(128, activation='relu'),
        k.layers.Dense(64, activation='relu'),
        k.layers.Dense(num_classes, activation='softmax')
    ])
    return model

def build_model_2(input_dim, num_classes, dropout_rate=0.2):
    """Now Building a deeper neural network with dropout regularization."""
    model = k.models.Sequential([
        k.layers.Dense(512, activation='relu', input_shape=(input_dim,)),
        k.layers.Dropout(dropout_rate),
        k.layers.Dense(256, activation='relu'),
        k.layers.Dropout(dropout_rate),
        k.layers.Dense(128, activation='relu'),
        k.layers.Dropout(dropout_rate),
        k.layers.Dense(64, activation='relu'),
        k.layers.Dropout(dropout_rate),
        k.layers.Dense(num_classes, activation='softmax')
    ])
    return model

def build_model_3(input_dim, num_classes, dropout_rate=0.2):
    """architecture for model 2 ; intended for training with a different optimizer."""
    return build_model_2(input_dim, num_classes, dropout_rate)

def build_model_4(input_dim, num_classes, dropout_rate=0.3):
    """Building the even deeper neural network with increased layer sizes and dropout."""
    model = k.models.Sequential([
        k.layers.Dense(1024, activation='relu', input_shape=(input_dim,)),
        k.layers.Dropout(dropout_rate),
        k.layers.Dense(512, activation='relu'),
        k.layers.Dropout(dropout_rate),
        k.layers.Dense(256, activation='relu'),
        k.layers.Dropout(dropout_rate),
        k.layers.Dense(128, activation='relu'),
        k.layers.Dropout(dropout_rate),
        k.layers.Dense(64, activation='relu'),
        k.layers.Dropout(dropout_rate),
        k.layers.Dense(num_classes, activation='softmax')
    ])
    return model

# =============================================================================
# MAIN EXECUTION FUNCTION
# =============================================================================
def main():
    # ---------------------
    # Data & Audio Visualization
    # ---------------------
    csv_path = '/content/drive/MyDrive/Data/features_3_sec.csv'
    df = load_dataset(csv_path)

    df_viz = df.copy()

    # Doing the Audio file demonstration
    audio_fp = '/content/drive/MyDrive/Data/genres_original/blues/blues.00000.wav'
    audio_data, sr = load_and_trim_audio(audio_fp)
    print("Playing sample audio...")
    display(play_audio(audio_data, sr))

    plot_waveform(audio_data, sr)
    plot_stft(audio_data, sr)
    plot_zoomed_wave(audio_data)
    plot_mel_spectrogram(audio_data, sr)
    plot_chroma(audio_data, sr)

    # Now Doing Feature correlation and tempo distribution plots
    plot_correlation_heatmap(df)
    plot_tempo_boxplot(df)

    # PCA visualization using features (and excluding the non-feature columns)
    features_viz = df_viz.drop(columns=["filename", "length", "label"])
    scaler_minmax = MinMaxScaler()
    X_scaled = pd.DataFrame(scaler_minmax.fit_transform(features_viz), columns=features_viz.columns)
    plot_pca(X_scaled, df_viz['label'])

    # ---------------------
    # Here I'm Doing Data Preprocessing for Modeling
    # ---------------------
    X_train, X_val, X_test, y_train, y_val, y_test, genre_to_index, index_to_genre = preprocess_data(df, seed)
    input_dim = X_train.shape[1]
    num_classes = 10  # Adjust if the number of genres changes

    print(f"Using TensorFlow version: {tf.__version__}")

    # ---------------------
    # Model Training & Evaluation
    # ---------------------
    # Model 1: Simple Feedforward Network
    print("\n--- Training Model 1 ---")
    model1 = build_model_1(input_dim, num_classes)
    model1.summary()
    history1 = train_and_plot(model1, X_train, y_train, X_val, y_val, epochs=70, optimizer='adam')

    # Model 2: Deep Neural Network with Dropout
    print("\n--- Training Model 2 ---")
    model2 = build_model_2(input_dim, num_classes)
    model2.summary()
    history2 = train_and_plot(model2, X_train, y_train, X_val, y_val, epochs=100, optimizer='adam')

    # Model 3: Same as Model 2 but here i'm using SGD and more epochs
    print("\n--- Training Model 3 ---")
    model3 = build_model_3(input_dim, num_classes)
    model3.summary()
    history3 = train_and_plot(model3, X_train, y_train, X_val, y_val, epochs=700, optimizer='sgd')

    # Model 4: Deeper Network with Increased Layer Sizes & Dropout (final model)
    print("\n--- Training Model 4 ---")
    model4 = build_model_4(input_dim, num_classes)
    model4.summary()
    history4 = train_and_plot(model4, X_train, y_train, X_val, y_val, epochs=500, optimizer='rmsprop')

    # Saving the final model
    model4.save("genre_classifier_model.h5")

    # Evaluate final model on the test dataset
    test_loss, test_accuracy = model4.evaluate(X_test, y_test, batch_size=128)
    print(f"Test Loss: {test_loss:.4f}")
    print(f"Test Accuracy: {test_accuracy * 100:.2f}%")


    # Predict class probabilities and then convert to class labels
    y_pred_prob = model4.predict(X_test, batch_size=128)
    y_pred = np.argmax(y_pred_prob, axis=1)

    # Create a sorted list of target names based on the index mapping
    target_names = [index_to_genre[i] for i in sorted(index_to_genre.keys())]

    # Printing the classification report
    report = skm.classification_report(y_test, y_pred, target_names=target_names)
    print("Classification Report:")
    print(report)

    # Compute and displaying the confusion matrix
    cm = skm.confusion_matrix(y_test, y_pred)
    plt.figure(figsize=(10, 8))
    sns.heatmap(cm, annot=True, fmt="d", cmap="viridis", xticklabels=target_names, yticklabels=target_names)
    plt.title("Confusion Matrix")
    plt.xlabel("Predicted")
    plt.ylabel("True")
    plt.tight_layout()
    plt.show()

    precision = skm.precision_score(y_test, y_pred, average="macro")
    recall = skm.recall_score(y_test, y_pred, average="macro")
    f1 = skm.f1_score(y_test, y_pred, average="macro")
    print(f"Macro Precision: {precision:.4f}")
    print(f"Macro Recall: {recall:.4f}")
    print(f"Macro F1 Score: {f1:.4f}")


if __name__ == "__main__":
    main()
```

```python
# =============================================================================
# END-TO-END INTEGRATED CODE FOR AUDIO CLASSIFICATION & FEATURE EXTRACTION
# =============================================================================

# =============================================================================
# IMPORT ALL NECESSARY LIBRARIES
# =============================================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_style('whitegrid')

import warnings
warnings.filterwarnings('ignore')

import sklearn.metrics as skm
import sklearn.model_selection as skms
import sklearn.preprocessing as skp
from sklearn.decomposition import PCA
from sklearn.preprocessing import MinMaxScaler

import librosa
import librosa.display as lplt
import IPython

import tensorflow as tf
import keras as k

# Set random seeds for reproducibility
seed = 12
np.random.seed(seed)
tf.random.set_seed(seed)

# libraries for transformer-based audio classification
import re
import gc
import time
from pathlib import Path
import torchaudio
from collections import Counter
from imblearn.over_sampling import RandomOverSampler
from datasets import Dataset
from tqdm.notebook import tqdm
tqdm.pandas()  # Enable progress tracking in DataFrame.apply
import evaluate
from transformers import (
    AutoFeatureExtractor, AutoModelForAudioClassification,
    TrainingArguments, Trainer, pipeline
)
from sklearn.metrics import roc_auc_score


def load_and_trim_audio(audio_fp):
    """Load an audio file and trim silent edges using librosa."""
    audio_data, sr = librosa.load(audio_fp)
    audio_data, _ = librosa.effects.trim(audio_data)
    return audio_data, sr

def play_audio(audio_data, sr):
    """Return an IPython Audio object for playback."""
    return IPython.display.Audio(audio_data, rate=sr)

def plot_waveform(audio_data, sr):
    """Plot the full audio waveform."""
    plt.figure(figsize=(15, 5))
    librosa.display.waveshow(audio_data, sr=sr)
    plt.xlabel("Time (s)")
    plt.ylabel("Amplitude")
    plt.title("Waveform")
    plt.show()

def plot_stft(audio_data, sr, n_fft=2048, hop_length=512, cmap='cool'):
    """Display a log-scaled short-time Fourier transform (STFT) spectrogram."""
    stft = librosa.stft(audio_data, n_fft=n_fft, hop_length=hop_length)
    stft_db = librosa.amplitude_to_db(np.abs(stft), ref=np.max)
    plt.figure(figsize=(12, 4))
    librosa.display.specshow(stft_db, sr=sr, hop_length=hop_length,
                             x_axis='time', y_axis='log', cmap=cmap)
    plt.colorbar(format='%+2.0f dB')
    plt.title("Log-scaled STFT Spectrogram (dB)")
    plt.tight_layout()
    plt.show()

def plot_zoomed_wave(audio_data, start=1000, end=1200):
    """Plot a zoomed segment of the audio waveform."""
    plt.figure(figsize=(16, 4))
    plt.plot(audio_data[start:end])
    plt.title("Zoomed Audio Waveform")
    plt.show()

def plot_mel_spectrogram(audio_data, sr, n_fft=2048, hop_length=512, n_mels=128, cmap='cool'):
    """Display the Mel spectrogram (in dB) of the audio."""
    mel_spec = librosa.feature.melspectrogram(y=audio_data, sr=sr, n_fft=n_fft,
                                              hop_length=hop_length, n_mels=n_mels)
    mel_spec_db = librosa.amplitude_to_db(mel_spec, ref=np.max)
    plt.figure(figsize=(16, 6))
    librosa.display.specshow(mel_spec_db, sr=sr, hop_length=hop_length,
                             x_axis='time', y_axis='mel', cmap=cmap)
    plt.colorbar(format='%+2.0f dB')
    plt.title("Mel Spectrogram (dB)")
    plt.tight_layout()
    plt.show()

def plot_chroma(audio_data, sr, hop_length=512, cmap='coolwarm'):
    """Plot chroma feature representation of the audio."""
    chroma = librosa.feature.chroma_stft(y=audio_data, sr=sr, hop_length=hop_length)
    plt.figure(figsize=(16, 6))
    librosa.display.specshow(chroma, sr=sr, hop_length=hop_length,
                             x_axis='time', y_axis='chroma', cmap=cmap)
    plt.colorbar()
    plt.title("Chroma Features")
    plt.tight_layout()
    plt.show()

# =============================================================================
# FEATURE ANALYSIS & VISUALIZATION FUNCTIONS (Keras/TensorFlow)
# =============================================================================

def load_dataset(csv_path):
    """Load CSV dataset and print basic summary info."""
    df = pd.read_csv(csv_path)
    print("Dataset has", df.shape)
    print("Count of samples by label:")
    print(df.label.value_counts().reset_index())
    return df

def plot_correlation_heatmap(df, save_path="mean_feature_correlation_heatmap.png"):
    """Plot a heatmap of mean-based feature correlations."""
    mean_cols = [col for col in df.columns if 'mean' in col]
    corr_matrix = df[mean_cols].corr()
    mask = np.triu(np.ones_like(corr_matrix, dtype=bool))
    plt.figure(figsize=(16, 11))
    custom_cmap = sns.diverging_palette(0, 25, s=90, l=45, n=5, as_cmap=True)
    sns.heatmap(corr_matrix, mask=mask, cmap=custom_cmap, vmax=0.3, center=0,
                square=True, linewidths=0.5, cbar_kws={"shrink": 0.5})
    plt.title("Feature Correlation Heatmap (Mean-based)", fontsize=20)
    plt.xticks(fontsize=10, rotation=45, ha='right')
    plt.yticks(fontsize=10)
    plt.tight_layout()
    plt.savefig(save_path)
    plt.show()

def plot_tempo_boxplot(df, save_path="genre_tempo_boxplot.png"):
    """Plot tempo distributions by genre using boxplots."""
    tempo_data = df[["label", "tempo"]]
    plt.figure(figsize=(16, 8))
    sns.boxplot(data=tempo_data, x="label", y="tempo", palette="husl")
    plt.title("Tempo Distribution by Genre (BPM)", fontsize=20)
    plt.xlabel("Music Genre", fontsize=15)
    plt.ylabel("Beats Per Minute", fontsize=15)
    plt.xticks(fontsize=14, rotation=45, ha='right')
    plt.yticks(fontsize=10)
    plt.tight_layout()
    plt.savefig(save_path)
    plt.show()

def plot_pca(X_normalized, genres, save_path="pca_genre_scatterplot.png"):
    """Perform PCA on normalized features and plot the first two principal components."""
    pca = PCA(n_components=2)
    pca_result = pca.fit_transform(X_normalized)
    pca_df = pd.DataFrame(pca_result, columns=["PC1", "PC2"])
    pca_df['Genre'] = genres
    plt.figure(figsize=(16, 9))
    sns.scatterplot(data=pca_df, x="PC1", y="PC2", hue="Genre", alpha=0.7, s=100)
    plt.title("PCA Projection of Music Genres", fontsize=20)
    plt.xlabel("Principal Component 1", fontsize=15)
    plt.ylabel("Principal Component 2", fontsize=15)
    plt.xticks(fontsize=14)
    plt.yticks(fontsize=10)
    plt.tight_layout()
    plt.savefig(save_path)
    plt.show()

# =============================================================================
# DATA PREPROCESSING FUNCTIONS (Keras/TensorFlow)
# =============================================================================

def preprocess_data(df, seed=12):
    """
    Process the dataset by mapping genre labels to numeric indices,
    shuffling and splitting into training, validation, and test sets.
    Also standardizes the features.
    """
    # Map genre labels to integers
    genre_to_index = {label: idx for idx, label in enumerate(df['label'].unique())}
    index_to_genre = {idx: label for label, idx in genre_to_index.items()}
    print("Genre to Index Mapping:\n", genre_to_index)
    print("\nIndex to Genre Mapping:\n", index_to_genre)

    # Replace string labels with integer indices
    df['label'] = df['label'].map(genre_to_index)

    # Shuffle dataset and remove non-feature columns
    df_shuffled = df.sample(frac=1, random_state=seed).reset_index(drop=True)
    df_shuffled.drop(columns=["filename", "length"], inplace=True)

    # Separate features and labels
    target = df_shuffled.pop("label")
    features = df_shuffled

    # Split into training (70%), validation (20%), and test (10%)
    X_train, X_temp, y_train, y_temp = skms.train_test_split(
        features, target, train_size=0.7, random_state=seed, stratify=target
    )
    X_val, X_test, y_val, y_test = skms.train_test_split(
        X_temp, y_temp, train_size=0.66, random_state=seed, stratify=y_temp
    )

    total = len(df_shuffled)
    print(f"Training Set: {X_train.shape[0]} samples ({round(X_train.shape[0] / total * 100)}% of total)")
    print(f"Validation Set: {X_val.shape[0]} samples ({round(X_val.shape[0] / total * 100)}% of total)")
    print(f"Test Set: {X_test.shape[0]} samples ({round(X_test.shape[0] / total * 100)}% of total)")

    print(f"Training Set - Majority Class Proportion: {y_train.value_counts(normalize=True).max() * 100:.2f}%")
    print(f"Validation Set - Majority Class Proportion: {y_val.value_counts(normalize=True).max() * 100:.2f}%")
    print(f"Test Set - Majority Class Proportion: {y_test.value_counts(normalize=True).max() * 100:.2f}%")

    # Standardize the features
    scaler = skp.StandardScaler()
    X_train_scaled = pd.DataFrame(scaler.fit_transform(X_train), columns=X_train.columns)
    X_val_scaled = pd.DataFrame(scaler.transform(X_val), columns=X_train.columns)
    X_test_scaled = pd.DataFrame(scaler.transform(X_test), columns=X_train.columns)

    return X_train_scaled, X_val_scaled, X_test_scaled, y_train, y_val, y_test, genre_to_index, index_to_genre

# =============================================================================
# MODEL TRAINING FUNCTIONS & ARCHITECTURES (Keras/TensorFlow)
# =============================================================================

ACCURACY_THRESHOLD = 0.94

class AccuracyEarlyStop(k.callbacks.Callback):
    """
    Custom early stopping callback to terminate training when validation accuracy exceeds a threshold.
    """
    def on_epoch_end(self, epoch, logs=None):
        if logs.get("val_accuracy") > ACCURACY_THRESHOLD:
            print(f"\nEarly stopping: Validation accuracy exceeded {ACCURACY_THRESHOLD * 100:.2f}%")
            self.model.stop_training = True

def train_and_plot(model, X_train, y_train, X_val, y_val, epochs, optimizer, batch_size=128):
    """Compile and train the model then plot the training history."""
    model.compile(optimizer=optimizer,
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])

    history = model.fit(
        X_train, y_train,
        validation_data=(X_val, y_val),
        epochs=epochs,
        batch_size=batch_size,
        callbacks=[AccuracyEarlyStop()]
    )

    best_val_acc = max(history.history["val_accuracy"])
    print(f"Highest Validation Accuracy Achieved: {best_val_acc:.4f}")

    pd.DataFrame(history.history).plot(figsize=(12, 6))
    plt.title("Training and Validation Metrics Over Epochs")
    plt.xlabel("Epoch")
    plt.ylabel("Metric Value")
    plt.grid(True)
    plt.tight_layout()
    plt.show()
    return history

def build_model_1(input_dim, num_classes):
    """Build a simple feedforward neural network."""
    model = k.models.Sequential([
        k.layers.Dense(256, activation='relu', input_shape=(input_dim,)),
        k.layers.Dense(128, activation='relu'),
        k.layers.Dense(64, activation='relu'),
        k.layers.Dense(num_classes, activation='softmax')
    ])
    return model

def build_model_2(input_dim, num_classes, dropout_rate=0.2):
    """Build a deep neural network with dropout regularization."""
    model = k.models.Sequential([
        k.layers.Dense(512, activation='relu', input_shape=(input_dim,)),
        k.layers.Dropout(dropout_rate),
        k.layers.Dense(256, activation='relu'),
        k.layers.Dropout(dropout_rate),
        k.layers.Dense(128, activation='relu'),
        k.layers.Dropout(dropout_rate),
        k.layers.Dense(64, activation='relu'),
        k.layers.Dropout(dropout_rate),
        k.layers.Dense(num_classes, activation='softmax')
    ])
    return model

def build_model_3(input_dim, num_classes, dropout_rate=0.2):
    """Alternate configuration using the same architecture as Model 2."""
    return build_model_2(input_dim, num_classes, dropout_rate)

def build_model_4(input_dim, num_classes, dropout_rate=0.3):
    """Build a deeper neural network with increased layer sizes and dropout."""
    model = k.models.Sequential([
        k.layers.Dense(1024, activation='relu', input_shape=(input_dim,)),
        k.layers.Dropout(dropout_rate),
        k.layers.Dense(512, activation='relu'),
        k.layers.Dropout(dropout_rate),
        k.layers.Dense(256, activation='relu'),
        k.layers.Dropout(dropout_rate),
        k.layers.Dense(128, activation='relu'),
        k.layers.Dropout(dropout_rate),
        k.layers.Dense(64, activation='relu'),
        k.layers.Dropout(dropout_rate),
        k.layers.Dense(num_classes, activation='softmax')
    ])
    return model

# =============================================================================
# MAIN FUNCTION FOR ORIGINAL Keras/TensorFlow AUDIO CLASSIFICATION
# =============================================================================

def main_original():
    """Main execution for original audio processing and genre classification."""
    # ---------- Data & Audio Visualization ----------
    csv_path = '/content/drive/MyDrive/Data/features_3_sec.csv'
    df = load_dataset(csv_path)
    df_viz = df.copy()

    # Audio demonstration on a sample file
    audio_fp = '/content/drive/MyDrive/Data/genres_original/blues/blues.00000.wav'
    audio_data, sr = load_and_trim_audio(audio_fp)
    print("Playing sample audio...")
    display(play_audio(audio_data, sr))

    plot_waveform(audio_data, sr)
    plot_stft(audio_data, sr)
    plot_zoomed_wave(audio_data)
    plot_mel_spectrogram(audio_data, sr)
    plot_chroma(audio_data, sr)

    # Feature correlation and tempo distributions
    plot_correlation_heatmap(df)
    plot_tempo_boxplot(df)

    # PCA visualization using normalized features
    features_viz = df_viz.drop(columns=["filename", "length", "label"])
    scaler_minmax = MinMaxScaler()
    X_scaled = pd.DataFrame(scaler_minmax.fit_transform(features_viz), columns=features_viz.columns)
    plot_pca(X_scaled, df_viz['label'])

    # ---------- Data Preprocessing for Modeling ----------
    X_train, X_val, X_test, y_train, y_val, y_test, genre_to_index, index_to_genre = preprocess_data(df, seed)
    input_dim = X_train.shape[1]
    num_classes = 10  # Adjust according to the number of genres

    print(f"Using TensorFlow version: {tf.__version__}")

    # ---------- Model Training & Evaluation ----------
    print("\n--- Training Model 1 ---")
    model1 = build_model_1(input_dim, num_classes)
    model1.summary()
    history1 = train_and_plot(model1, X_train, y_train, X_val, y_val, epochs=70, optimizer='adam')

    print("\n--- Training Model 2 ---")
    model2 = build_model_2(input_dim, num_classes)
    model2.summary()
    history2 = train_and_plot(model2, X_train, y_train, X_val, y_val, epochs=100, optimizer='adam')

    print("\n--- Training Model 3 ---")
    model3 = build_model_3(input_dim, num_classes)
    model3.summary()
    history3 = train_and_plot(model3, X_train, y_train, X_val, y_val, epochs=700, optimizer='sgd')

    print("\n--- Training Model 4 ---")
    model4 = build_model_4(input_dim, num_classes)
    model4.summary()
    history4 = train_and_plot(model4, X_train, y_train, X_val, y_val, epochs=500, optimizer='rmsprop')

    # Save final model and evaluate on test set
    model4.save("genre_classifier_model.h5")
    test_loss, test_accuracy = model4.evaluate(X_test, y_test, batch_size=128)
    print(f"Test Loss: {test_loss:.4f}")
    print(f"Test Accuracy: {test_accuracy * 100:.2f}%")

    # Generate classification report and confusion matrix
    y_pred_prob = model4.predict(X_test, batch_size=128)
    y_pred = np.argmax(y_pred_prob, axis=1)

    target_names = [index_to_genre[i] for i in sorted(index_to_genre.keys())]
    report = skm.classification_report(y_test, y_pred, target_names=target_names)
    print("Classification Report:")
    print(report)

    cm = skm.confusion_matrix(y_test, y_pred)
    plt.figure(figsize=(10, 8))
    sns.heatmap(cm, annot=True, fmt="d", cmap="viridis", xticklabels=target_names, yticklabels=target_names)
    plt.title("Confusion Matrix")
    plt.xlabel("Predicted")
    plt.ylabel("True")
    plt.tight_layout()
    plt.show()

    precision = skm.precision_score(y_test, y_pred, average="macro")
    recall = skm.recall_score(y_test, y_pred, average="macro")
    f1 = skm.f1_score(y_test, y_pred, average="macro")
    print(f"Macro Precision: {precision:.4f}")
    print(f"Macro Recall: {recall:.4f}")
    print(f"Macro F1 Score: {f1:.4f}")

# =============================================================================
#  AUDIO CLASSIFICATION USING TRANSFORMERS & TORCHAUDIO
# =============================================================================

# Define constants and label mappings for transformer workflow
RESAMPLE_RATE = 16000       # Resampling rate in Hz
MAX_AUDIO_LEN = 480000      # Maximum audio length (samples)
genre_labels = ['disco', 'metal', 'reggae', 'blues', 'rock',
                'classical', 'jazz', 'hiphop', 'country', 'pop']

label_to_id = {label: idx for idx, label in enumerate(genre_labels)}
id_to_label = {idx: label for label, idx in label_to_id.items()}
print("id_to_label mapping:", id_to_label)
print("label_to_id mapping:", label_to_id)

def retrieve_audio_data():
    """
    Scan the specified folder for .wav files (excluding problematic samples),
    extract genre from filename, and return a DataFrame with file paths and labels.
    """
    files_list = []
    labels_list = []
    base_path = Path('/content/drive/MyDrive/Data/genres_original/')
    for file_path in base_path.glob('*/*.wav'):
        if 'jazz.00054' not in str(file_path):
            match_obj = re.match(r"(\D+)", file_path.stem)
            if match_obj:
                genre_name = match_obj.group(1).rstrip('.')
                files_list.append(file_path)
                labels_list.append(label_to_id[genre_name])
    data_df = pd.DataFrame({'file': files_list, 'label': labels_list})
    return data_df

def transform_audio_sample(file_path):
    """
    Load, resample, and truncate an audio file.
    Returns a NumPy array of audio samples.
    """
    audio_tensor, orig_rate = torchaudio.load(str(file_path))
    resample_transform = torchaudio.transforms.Resample(orig_freq=orig_rate, new_freq=RESAMPLE_RATE)
    audio_resampled = resample_transform(audio_tensor).squeeze(0).numpy()
    return audio_resampled[:MAX_AUDIO_LEN]

def main_transformer():
    """Main execution for transformer-based audio classification."""
    start_time = time.time()
    # ---- Data Loading & Preprocessing ----
    audio_df = retrieve_audio_data()
    print("Audio data loaded. Sample records:")
    print(audio_df.head())
    print("Load time: {:.2f} seconds".format(time.time() - start_time))

    # Balance dataset via oversampling
    y_vals = audio_df[['label']]
    audio_df_nolab = audio_df.drop(['label'], axis=1)
    oversampler = RandomOverSampler(random_state=83)
    balanced_files, balanced_labels = oversampler.fit_resample(audio_df_nolab, y_vals)
    audio_df = balanced_files.copy()
    audio_df['label'] = balanced_labels
    del y_vals, balanced_files, balanced_labels
    gc.collect()

    print("Dataset shape after oversampling:", audio_df.shape)
    print("Label counts after balancing:")
    print(audio_df['label'].value_counts())

    # ---- Audio Transformation ----
    print("Applying audio transformation (this may take a while)...")
    audio_df['audio'] = audio_df['file'].progress_apply(transform_audio_sample)

    # Drop the raw file paths column
    drop_start = time.time()
    audio_df = audio_df.drop(['file'], axis=1)
    print("Dropped file paths. Time taken: {:.2f} seconds".format(time.time() - drop_start))
    print("Sample transformed records:")
    print(audio_df.sample(5))

    # ---- Check for and Remove Records with NaNs in Audio ----
    audio_df['has_nan'] = audio_df['audio'].apply(lambda a: np.isnan(a).any())
    print("Number of records with NaN in audio:", audio_df['has_nan'].sum())
    audio_df = audio_df[~audio_df['has_nan']].drop(columns=['has_nan'])

    # ---- Convert to Hugging Face Dataset and Split ----
    hf_dataset = Dataset.from_pandas(audio_df)
    label_freq = Counter(hf_dataset['label'])
    print("Label frequency in dataset:", list(label_freq.items()))

    hf_splits = hf_dataset.train_test_split(test_size=0.2)
    print("Dataset splits:")
    print(hf_splits)

    # ---- Load Feature Extractor & Pre-trained Model ----
    model_identifier = "facebook/wav2vec2-base-960h"
    feat_extractor = AutoFeatureExtractor.from_pretrained(model_identifier)
    hf_model = AutoModelForAudioClassification.from_pretrained(model_identifier, num_labels=len(genre_labels))
    hf_model.config.id2label = id_to_label

    print("Trainable parameters (M):", hf_model.num_parameters(only_trainable=True) / 1e6)

    # ---- Preprocess Dataset ----
    def preprocess_batch(batch):
        inputs = feat_extractor(batch['audio'], sampling_rate=RESAMPLE_RATE,
                                max_length=MAX_AUDIO_LEN, truncation=True)
        inputs['input_values'] = inputs['input_values'][0]
        return inputs

    hf_splits['test'] = hf_splits['test'].map(preprocess_batch, remove_columns="audio", batched=False)
    hf_splits['train'] = hf_splits['train'].map(preprocess_batch, remove_columns="audio", batched=False)
    gc.collect()

    # ---- Define Metrics for Evaluation ----
    accuracy_metric = evaluate.load("accuracy")
    def compute_metrics(eval_pred):
        logits = eval_pred.predictions
        exp_logits = np.exp(logits)
        probs = exp_logits / np.sum(exp_logits, axis=1, keepdims=True)
        labels_true = eval_pred.label_ids
        roc_auc = roc_auc_score(labels_true, probs, average='macro', multi_class='ovr')
        accuracy = accuracy_metric.compute(predictions=probs.argmax(axis=1), references=labels_true)['accuracy']
        return {"roc_auc": roc_auc, "accuracy": accuracy}

    # ---- Training Arguments & Trainer Initialization ----
    train_args = TrainingArguments(
        output_dir="music_genres_classification",
        logging_dir='./logs',
        num_train_epochs=25,
        per_device_train_batch_size=4,
        per_device_eval_batch_size=4,
        learning_rate=1e-5,
        logging_strategy='steps',
        logging_first_step=True,
        load_best_model_at_end=True,
        logging_steps=1,
        eval_strategy='epoch',
        warmup_steps=50,
        weight_decay=0.02,
        eval_steps=1,
        gradient_accumulation_steps=1,
        gradient_checkpointing=True,
        save_strategy='epoch',
        save_total_limit=1,
        report_to=[],
    )

    audio_trainer = Trainer(
        model=hf_model,
        args=train_args,
        train_dataset=hf_splits["train"],
        eval_dataset=hf_splits["test"],
        tokenizer=feat_extractor,
        compute_metrics=compute_metrics,
    )

    # ---- Baseline Evaluation ----
    print("Baseline evaluation:")
    baseline_results = audio_trainer.evaluate()
    print(baseline_results)

    # ---- Train Transformer-based Model ----
    print("Starting training...")
    audio_trainer.train()

    print("Evaluation after training:")
    final_eval = audio_trainer.evaluate()
    print(final_eval)

    audio_trainer.save_model()

    # ---- Set Up Inference Pipeline ----
    audio_pipe = pipeline('audio-classification', model="music_genres_classification", device=0)

    # ---- Example Inference: "disco" sample ----
    audio_sample_disco, rate_disco = torchaudio.load('/content/drive/MyDrive/Data/genres_original/disco/disco.00005.wav')
    resample_disco = torchaudio.transforms.Resample(orig_freq=rate_disco, new_freq=RESAMPLE_RATE)
    audio_disco = resample_disco(audio_sample_disco).numpy().reshape(-1)
    print("Inference for a disco sample:")
    print(audio_pipe(audio_disco))

    # ---- Example Inference: "classical" sample ----
    audio_sample_class, rate_class = torchaudio.load('/content/drive/MyDrive/Data/genres_original/classical/classical.00014.wav')
    resample_class = torchaudio.transforms.Resample(orig_freq=rate_class, new_freq=RESAMPLE_RATE)
    audio_class = resample_class(audio_sample_class).numpy().reshape(-1)
    print("Inference for a classical sample:")
    print(audio_pipe(audio_class))

    from IPython.display import Audio
    display(Audio(audio_class, rate=RESAMPLE_RATE))

# =============================================================================
# ENTRY POINT
# =============================================================================

if __name__ == "__main__":
  # main_original()

  main_transformer()

```
